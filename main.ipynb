```python
import sympy as sp

# 1. Задаем переменные
x, y, z, u = sp.symbols('x y z u')

# 2. Записываем уравнения
eq1 = sp.Eq(32*x + 97*y + 90*z + 48*u, 47)
eq2 = sp.Eq(35*x + 84*y + 91*z + 54*u, 40)
eq3 = sp.Eq(17*x + 18*y + 21*z + 13*u, 37)

# 3. Решаем систему. 
# Мы говорим solve, что хотим выразить y, z, u через x.
# (x остается свободной переменной)
solution = sp.solve((eq1, eq2, eq3), (y, z, u))

# 4. Выводим решение для y
y_expr = solution[y]
print("y =", y_expr)

# 5. Раскрываем скобки и собираем коэффициенты при x, чтобы найти a и b
y_expanded = sp.expand(y_expr)
print("Раскрытый вид y =", y_expanded)

# Извлекаем константу a (свободный член) и b (коэффициент при x)
a = y_expanded.subs(x, 0)  # подставляем x=0, остается только константа
b = y_expanded.coeff(x)    # Берем коэффициент при x

print(f"\nКонстанта a = {a}")
print(f"Константа b = {b}")
```

```python
import sympy as sp

def det_min(n):
    """Определитель матрицы a_ij = min(i, j)"""
    A = sp.Matrix(n, n, lambda i, j: min(i+1, j+1))
    return sp.simplify(A.det())

def det_max(n):
    """Определитель матрицы a_ij = max(i, j)"""
    A = sp.Matrix(n, n, lambda i, j: max(i+1, j+1))
    return sp.simplify(A.det())

def det_abs_diff(n):
    """Определитель матрицы a_ij = |i - j|"""
    A = sp.Matrix(n, n, lambda i, j: abs(i - j))
    return sp.simplify(A.det())

print("1.5. Определители порядка n")
for n in range(1, 8):
    print(f"n={n}:")
    print(f"  a) min(i,j):   det = {det_min(n)}")
    print(f"  б) max(i,j):   det = {det_max(n)}")
    print(f"  в) |i-j|:      det = {det_abs_diff(n)}")


n = sp.symbols('n', positive=True, integer=True)

# a)
print("\nОбщая формула для a): det = (-1)^(n-1) * n")
for n_val in range(1, 9):
    formula = (-1)**(n_val-1) * n_val
    actual = det_min(n_val)
    print(f"n={n_val}: formula={formula}, actual={actual}, ok={formula==actual}")

# б) max(i,j)
print("\nОбщая формула для б): det = (-1)^(n-1) * n")
for n_val in range(1, 9):
    formula = (-1)**(n_val-1) * n_val
    actual = det_max(n_val)
    print(f"n={n_val}: formula={formula}, actual={actual}, ok={formula==actual}")

# в) |i-j|
print("\nОбщая формула для в): det = (-1)^(n-1) * 2^(n-2) * (n-1)")
for n_val in range(2, 9):
    formula = (-1)**(n_val-1) * 2**(n_val-2) * (n_val-1)
    actual = det_abs_diff(n_val)
    print(f"n={n_val}: formula={formula}, actual={actual}, ok={formula==actual}")
```

```python
n = 80000
a11 = 0.99
ann = 1.01

d = (ann - a11) / (n - 1)
print(f"1.7. n = {n}")
print(f"Шаг арифметической прогрессии d = {d}")

a22 = a11 + d
print(f"a11 = {a11}, a22 = {a22}")

det_2x2 = a11 * a22 - 0.57 * 0.03
print(f"det 2x2 блока = {det_2x2}")


prod = 1.0

import math
log_prod = 0.0
for k in range(3, n+1):
    log_prod += math.log(a11 + (k-1)*d)
prod = math.exp(log_prod)
print(f"Произведение a33...ann = {prod}")

det_A = det_2x2 * prod
print(f"det(A) = {det_A}")


inv_2x2 = np.array([[a11, 0.57],[0.03, a22]])
inv_2x2 = np.linalg.inv(inv_2x2)
tr_2x2 = np.trace(inv_2x2)
print(f"tr(2x2^{-1}) = {tr_2x2}")

sum_inv = 0.0
for k in range(3, n+1):
    sum_inv += 1.0 / (a11 + (k-1)*d)
print(f"sum 1/a_k (k=3..n) = {sum_inv}")

tr_inv = tr_2x2 + sum_inv
print(f"tr(A^{{-1}}) = {tr_inv}")

print(f"\nОтвет задачи 1.7:\n 1) det = {det_A:.6f}")
print(f" 2) tr = {tr_inv:.6f}")
```

элементом a1,1 = 1 и конечным элементом an,n = 1,07; 2) элементы первой строки образуют арифметическую прогрессию с конечным элементом a1,n = 1,10; 3) все элементы матрицы A, расположенные вне главной
диагонали и вне первой строки, равны нулю. Найдите обратную матрицу
B = A−1 и укажите в ответе: 1) наименьший элемент матрицы B; 2) сумму
элементов матрицы B.
Ответ: 1) −1, 028; 2) −2054, 4.

```python
import sympy as sp

n = 43000
a_11 = 1
a_1n = 1.10
a_nn = 1.07


q = a_nn**(1/(n-1))

d = (a_1n - a_11)/(n-1)


def b_1j(j):
    a_1j = a_11 + (j-1)*d
    d_j = q**(j-1)
    return -a_1j / d_j


min_val = 0
min_j = 0
for j in range(2, n+1):
    val = b_1j(j)
    if val < min_val:
        min_val = val
        min_j = j

print(f"Наименьший элемент B: {min_val:.4f} (при j={min_j})")


sum_diag = 1 + sum(1/(q**(j-1)) for j in range(2, n+1))
sum_first_row = sum(b_1j(j) for j in range(2, n+1))
total_sum = sum_diag + sum_first_row

print(f"Сумма элементов B: {total_sum:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def hadamard(n):
    """Матрица Адамара порядка n = 2^k (конструкция Сильвестра)"""
    if n == 1:
        return np.array([[1]])
    H = np.array([[1, 1], [1, -1]])
    while H.shape[0] < n:
        H = np.kron(H, np.array([[1, 1], [1, -1]]))
    return H

print("H1:\n", hadamard(1))
print("H2:\n", hadamard(2))
print("H4:\n", hadamard(4))
print("H2 ⊗ H2 == H4:", np.array_equal(np.kron(hadamard(2), hadamard(2)), hadamard(4)))

H64 = hadamard(64)

cmap = plt.cm.colors.ListedColormap(['#1E90FF', '#FF8C00'])
plot_data = (H64 + 1) // 2 
fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(plot_data, cmap=cmap, aspect='equal', origin='upper', interpolation='nearest')

ticks = [0, 19, 39, 63]
labels = ['1', '20', '40', '64']
ax.set_xticks(ticks)
ax.set_xticklabels(labels)
ax.set_yticks(ticks)
ax.set_yticklabels(labels)
ax.tick_params(labeltop=True, labelright=True)

ax.set_title(r'$H_{2^{6}}$', fontsize=16, pad=20)
plt.tight_layout()
plt.savefig('hadamard_64.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import sympy as sp

def solve_conic(a, b, c, d, e, f):

    print(f"\n--- Решение для уравнения: {a}x^2 + {2*b}xy + {c}y^2 + {2*d}x + {2*e}y + {f} = 0 ---")

    A = np.array([[a, b],
                  [b, c]])

    eigenvalues, eigenvectors = np.linalg.eigh(A)
    

    idx = eigenvalues.argsort()[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]

    print(f"Собственные значения: lambda1 = {eigenvalues[0]:.4f}, lambda2 = {eigenvalues[1]:.4f}")
    print(f"Собственные векторы (столбцы матрицы поворота):\n{eigenvectors}")

    d_vec = np.array([d, e])
    center = -np.linalg.inv(A) @ d_vec
    x0, y0 = center
    print(f"Центр кривой (x0, y0): ({x0:.4f}, {y0:.4f})")


    F_new = f - d_vec.T @ np.linalg.inv(A) @ d_vec
    print(f"Свободный член после переноса: F' = {F_new:.4f}")


    print("\nКаноническое уравнение:")
    if abs(F_new) < 1e-9:
        print(f"{eigenvalues[0]:.4f} * x'^2 + {eigenvalues[1]:.4f} * y'^2 = 0")
    else:
        A_sq = -F_new / eigenvalues[0]
        B_sq = -F_new / eigenvalues[1]
        if A_sq > 0 and B_sq > 0:
            print(f"x'^2 / {A_sq:.4f} + y'^2 / {B_sq:.4f} = 1")
        elif A_sq > 0 > B_sq:
            print(f"x'^2 / {A_sq:.4f} - y'^2 / {abs(B_sq):.4f} = 1")
        elif B_sq > 0 > A_sq:
            print(f"y'^2 / {B_sq:.4f} - x'^2 / {abs(A_sq):.4f} = 1")
        else:
            print(f"{eigenvalues[0]:.4f} * x'^2 + {eigenvalues[1]:.4f} * y'^2 + {F_new:.4f} = 0")
    print("\nКаноническая система координат:")
    print(f"x = {x0:.4f} + ({eigenvectors[0,0]:.4f}) * x' + ({eigenvectors[0,1]:.4f}) * y'")
    print(f"y = {y0:.4f} + ({eigenvectors[1,0]:.4f}) * x' + ({eigenvectors[1,1]:.4f}) * y'")

#а)

solve_conic(5, 2, 8, -16, -28, 80)

#б)

solve_conic(5, 6, 0, -11, -6, -19)
```

```python
import numpy as np
import sympy as sp

A1 = sp.Matrix([
    [2, 4, -2],
    [4, 9, -5],
    [-2, -5, 3]
])

A2 = sp.Matrix([
    [2, -2, -2],
    [-2, 3, 4],
    [-2, 4, 6]
])

def analyze_form(A, name):
    print(f"--- Анализ формы {name} ---")
    print(f"Матрица:\n{A}")
    
    # Собственные значения
    eigenvals = A.eigenvals()
    print(f"Собственные значения: {eigenvals}")
    
    # Ранг
    rank = A.rank()
    print(f"Ранг матрицы: {rank}")
    
    # Сигнатура (количество положительных и отрицательных собственных значений)
    pos = sum(1 for val in eigenvals for _ in range(eigenvals[val]) if val > 0)
    neg = sum(1 for val in eigenvals for _ in range(eigenvals[val]) if val < 0)
    print(f"Сигнатура: {pos} положительных, {neg} отрицательных")
    print(f"Нормальный вид: {pos} положительных квадратов, {neg} отрицательных квадратов\n")

analyze_form(A1, "Q1")
analyze_form(A2, "Q2")
```
